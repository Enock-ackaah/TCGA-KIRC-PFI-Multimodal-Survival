```python
def extract_gene_tpm(file_path):
    df = pd.read_csv(
        file_path,
        sep="\t",
        comment="#"
    )

    # Keep actual gene rows only
    df = df[df["gene_id"].astype(str).str.startswith("ENSG")].copy()

    # Keep gene identifier and TPM
    df = df[["gene_id", "gene_name", "tpm_unstranded"]].copy()

    return df
```

```python
comparison_results = []

for (case_id, sample_id), group in same_sample_profiles.groupby(
    ["Case ID", "Sample ID"]
):

    if len(group) != 2:
        continue

    row1 = group.iloc[0]
    row2 = group.iloc[1]

    path1 = find_expression_file(
        row1["File ID"],
        row1["File Name"]
    )

    path2 = find_expression_file(
        row2["File ID"],
        row2["File Name"]
    )

    df1 = extract_gene_tpm(path1)
    df2 = extract_gene_tpm(path2)

    merged = df1.merge(
        df2,
        on="gene_id",
        suffixes=("_file1", "_file2")
    )

    pearson_r = merged[
        ["tpm_unstranded_file1", "tpm_unstranded_file2"]
    ].corr(method="pearson").iloc[0, 1]

    spearman_r = merged[
        ["tpm_unstranded_file1", "tpm_unstranded_file2"]
    ].corr(method="spearman").iloc[0, 1]

    mean_abs_difference = np.mean(
        np.abs(
            merged["tpm_unstranded_file1"]
            - merged["tpm_unstranded_file2"]
        )
    )

    comparison_results.append(
        {
            "Case ID": case_id,
            "Sample ID": sample_id,
            "File 1": row1["File ID"],
            "File 2": row2["File ID"],
            "Genes compared": len(merged),
            "Pearson r": pearson_r,
            "Spearman rho": spearman_r,
            "Mean absolute TPM difference": mean_abs_difference
        }
    )

duplicate_expression_comparison = pd.DataFrame(
    comparison_results
)

display(duplicate_expression_comparison)
```

```python
log_comparison_results = []

for (case_id, sample_id), group in same_sample_profiles.groupby(
    ["Case ID", "Sample ID"]
):

    if len(group) != 2:
        continue

    row1 = group.iloc[0]
    row2 = group.iloc[1]

    path1 = find_expression_file(
        row1["File ID"],
        row1["File Name"]
    )

    path2 = find_expression_file(
        row2["File ID"],
        row2["File Name"]
    )

    df1 = extract_gene_tpm(path1)
    df2 = extract_gene_tpm(path2)

    merged = df1.merge(
        df2,
        on="gene_id",
        suffixes=("_file1", "_file2")
    )

    x = np.log2(merged["tpm_unstranded_file1"] + 1)
    y = np.log2(merged["tpm_unstranded_file2"] + 1)

    pearson_log = x.corr(y, method="pearson")
    spearman_log = x.corr(y, method="spearman")

    median_abs_log_difference = np.median(
        np.abs(x - y)
    )

    log_comparison_results.append(
        {
            "Case ID": case_id,
            "Sample ID": sample_id,
            "Genes compared": len(merged),
            "Log2(TPM+1) Pearson r": pearson_log,
            "Log2(TPM+1) Spearman rho": spearman_log,
            "Median absolute log2 difference":
                median_abs_log_difference
        }
    )

log_duplicate_comparison = pd.DataFrame(
    log_comparison_results
)

display(log_duplicate_comparison)
```

```python
url = "https://api.gdc.cancer.gov/cases"

filters = {
    "op": "=",
    "content": {
        "field": "project.project_id",
        "value": "TCGA-KIRC"
    }
}

fields = [
    "case_id",
    "submitter_id",
    "project.project_id",
    "demographic.age_at_index",
    "demographic.gender",
    "demographic.race",
    "demographic.ethnicity",
    "demographic.vital_status",
    "demographic.days_to_death",
    "diagnoses.age_at_diagnosis",
    "diagnoses.primary_diagnosis",
    "diagnoses.tumor_grade",
    "diagnoses.ajcc_pathologic_stage",
    "diagnoses.ajcc_pathologic_t",
    "diagnoses.ajcc_pathologic_n",
    "diagnoses.ajcc_pathologic_m",
    "diagnoses.days_to_last_follow_up"
]

params = {
    "filters": json.dumps(filters),
    "fields": ",".join(fields),
    "format": "JSON",
    "size": "1000"
}

response = requests.get(url, params=params, timeout=60)

print("HTTP status:", response.status_code)

response.raise_for_status()

gdc_cases = response.json()["data"]["hits"]

print("TCGA-KIRC clinical cases retrieved:", len(gdc_cases))
```

```python
def select_kirc_diagnosis(diagnoses):
    if not diagnoses:
        return {}

    # Prefer the renal clear-cell carcinoma record
    clear_cell = [
        d for d in diagnoses
        if "clear cell" in str(
            d.get("primary_diagnosis", "")
        ).lower()
    ]

    if clear_cell:
        # If more than one clear-cell record exists,
        # prefer one containing pathologic stage information
        staged = [
            d for d in clear_cell
            if d.get("ajcc_pathologic_stage") is not None
        ]

        if staged:
            return staged[0]

        return clear_cell[0]

    # Fallback only when no clear-cell diagnosis is available
    return diagnoses[0]
```

```python
# ------------------------------------------------------------
# Build survival endpoint datasets
# ------------------------------------------------------------

pfi_cohort = survival_matched[
    survival_matched["PFI"].notna()
    & survival_matched["PFI.time"].notna()
].copy()

os_cohort = survival_matched[
    survival_matched["OS"].notna()
    & survival_matched["OS.time"].notna()
].copy()

print("PFI cohort")
print("-----------")
print("Patients:", len(pfi_cohort))
print("Events:", int(pfi_cohort["PFI"].sum()))
print("Censored:", int((pfi_cohort["PFI"] == 0).sum()))
print("Minimum follow-up time:", pfi_cohort["PFI.time"].min())
print("Median follow-up/event time:", pfi_cohort["PFI.time"].median())
print("Maximum follow-up time:", pfi_cohort["PFI.time"].max())

print("\nOS cohort")
print("---------")
print("Patients:", len(os_cohort))
print("Deaths:", int(os_cohort["OS"].sum()))
print("Censored:", int((os_cohort["OS"] == 0).sum()))
print("Minimum follow-up time:", os_cohort["OS.time"].min())
print("Median follow-up/event time:", os_cohort["OS.time"].median())
print("Maximum follow-up time:", os_cohort["OS.time"].max())
```

```python
# ============================================================
# FINAL SURVIVAL ELIGIBILITY RULE
# Require:
#   1. nonmissing event indicator
#   2. nonmissing survival time
#   3. strictly positive survival time
# ============================================================

pfi_final = survival_matched[
    survival_matched["PFI"].notna()
    & survival_matched["PFI.time"].notna()
    & (survival_matched["PFI.time"] > 0)
].copy()

os_final = survival_matched[
    survival_matched["OS"].notna()
    & survival_matched["OS.time"].notna()
    & (survival_matched["OS.time"] > 0)
].copy()

print("FINAL PFI COHORT")
print("----------------")
print("Patients:", len(pfi_final))
print("Events:", int(pfi_final["PFI"].sum()))
print("Censored:", int((pfi_final["PFI"] == 0).sum()))
print("Minimum time:", pfi_final["PFI.time"].min())
print("Median time:", pfi_final["PFI.time"].median())
print("Maximum time:", pfi_final["PFI.time"].max())

print("\nFINAL OS COHORT")
print("---------------")
print("Patients:", len(os_final))
print("Deaths:", int(os_final["OS"].sum()))
print("Censored:", int((os_final["OS"] == 0).sum()))
print("Minimum time:", os_final["OS.time"].min())
print("Median time:", os_final["OS.time"].median())
print("Maximum time:", os_final["OS.time"].max())
```
